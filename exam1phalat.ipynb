```python
import math
def factor(n):
    factors = []
    for i in range(2,math.floor(math.sqrt(n))+1):
        while n % i == 0:
            factors.append(i)
            n //= i
    if n > 1:
        factors.append(n)
    return factors
```

```python
factor(360)
```

```python
from itertools import permutations
perm_num = permutations(range(1,11))
counter=0
for i in perm_num:
    R=i[0]
    U=i[1]
    L=i[2]
    E=i[3]
    T=i[4]
    O=i[5]
    P=i[6]
    N=i[7]
    if U==R+T and P==R+U and O==L+E and N==E+O :
        print(f"R={R}, U={U}, L={L}, E={E}, T={T}, O={O}, P={P}, N={N}")
        counter+=1
print(f"รูปแบบทั้งหมด: {counter} รูปแบบ")
```
