```python
D=int(input("D: "))
T=int(input("T: "))
H=int(input("H: "))
R=int(input("R: "))
B=int(input("B: "))
H_back=int(input("H_back: "))
Rabbit=0
Turtle=0

x=True
i=1
print("Time\tTurtle\tRabbit\tEvent")
print("-"*30)
while x:
    if i%R==0 and i%D==0:
        Rabbit+=0
        Event = "Rabbit Rest"
    elif i%R==0:
        Rabbit+=0
        Event = "Rabbit Rest"
    elif i%B==0:
        Rabbit=Rabbit-H_back
        Event = "Rabbit turn back"
        if Rabbit<0:
            Rabbit=0
    else:
        Rabbit+=H
        Event = "Rabbit runs"
    Turtle+=T
    print(f"{i}\t{Turtle}\t{Rabbit}\t{Event}")
    i+=1
    if Turtle >= D or Rabbit >= D:
        x=False
        break
print("-"*30)
print("Result: ",end="")
if Turtle > Rabbit:
    print("Turtle wins")
elif Rabbit > Turtle:
    print("Rabbit wins")
print("Finish Time: ",i-1,"seconds")
```
